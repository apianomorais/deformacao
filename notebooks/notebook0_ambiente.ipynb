```python
import sys
import numpy as np
import scipy
import matplotlib
from scipy.linalg import eigh
from scipy.optimize import brentq

print("Python     ", sys.version.split()[0])
print("NumPy      ", np.__version__)
print("SciPy      ", scipy.__version__)
print("Matplotlib ", matplotlib.__version__)

# um teste de verdade: o autovalor de uma matriz 2x2 conhecida
K = np.array([[2.0, -1.0], [-1.0, 2.0]])
M = np.eye(2)
lam, _ = eigh(K, M)
assert np.allclose(lam, [1.0, 3.0]), "algo errado com o SciPy"

print("ambiente OK")
```

## Som

No notebook (Jupyter ou Colab), o som sai com `IPython.display.Audio`.
No Spyder, use `sounddevice` (`pip install sounddevice`).

```python
from IPython.display import Audio
import numpy as np

fs = 44100
t = np.arange(0, 1.0, 1/fs)
Audio(0.3*np.sin(2*np.pi*440*t), rate=fs, autoplay=True)
```

## Gráfico

Se a figura aparecer, o matplotlib está configurado.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 1, 200)
plt.figure(figsize=(7, 3))
plt.plot(x, np.sin(2*np.pi*x))
plt.title("se você está vendo isto, está tudo certo")
plt.tight_layout()
plt.show()
```
